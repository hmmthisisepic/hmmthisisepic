## Write a Function to Calculate an Average
Write a function that accepts a list of numbers and returns their average. Do this by writing code to add up the numbers and divide by how many there are (don't use any built in functions).

Then use it to calculate the average height of all tournament winners by passing the data from the column called `winner_ht`.

Note that some of the height values are empty, so your average function must spot those and ignore them.

The answer should be 184.80318822808093

```python
def averagecalc(numbers):
  sum = 0
  count = 0
  for num in numbers: #make an average calculator
    sum += num
    count += 1
  if count == 0:
    return 0
  return sum / count

#test site bc requested
testing = [1, 2, 3]
testavgres = averagecalc(testing)
print("Test for average by 123")
print(testavgres)

rowselect = -1 #getting the winner_ht
for i in range(len(columnheading)): #continiously search for smth
  if columnheading[i] == 'winner_ht': #identify heading with this specific column
    rowselect = i
    break #end search

wutheringheights = []
for row in datatable:
  howtall = row[rowselect] #turns out this thing cannot select the thing on its own wow. 1 moment
  if howtall != '': #this thing filters out empty values
    wutheringheights.append(float(howtall))

avgwintall = averagecalc(wutheringheights)
print("Average winner heights:")
print(avgwintall)
```
